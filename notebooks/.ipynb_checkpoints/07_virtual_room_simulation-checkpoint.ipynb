# 07 Virtual Room Simulation

2D room visualization with transmitter, receiver, zone layout, and moving person silhouette.

This notebook is the bridge from research notebooks to the eventual React/Three.js dashboard.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Rectangle

room = {"width_m": 5.0, "depth_m": 4.0}
tx = np.array([0.6, 2.0])
rx = np.array([4.4, 2.0])

n_frames = 8
person_path = np.column_stack([
    np.linspace(1.0, 4.0, n_frames),
    1.15 + 0.7 * np.sin(np.linspace(0, 2 * np.pi, n_frames)),
])

zones = {
    "left": (0.0, 0.0, room["width_m"] / 3, room["depth_m"]),
    "center": (room["width_m"] / 3, 0.0, room["width_m"] / 3, room["depth_m"]),
    "right": (2 * room["width_m"] / 3, 0.0, room["width_m"] / 3, room["depth_m"]),
}
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
colors = {"left": "#d9edf7", "center": "#e8f5e9", "right": "#fff3cd"}
for name, (x, y, w, h) in zones.items():
    ax.add_patch(Rectangle((x, y), w, h, facecolor=colors[name], edgecolor="white", alpha=0.85))
    ax.text(x + w / 2, y + h - 0.25, name, ha="center", va="center", fontsize=11)

ax.plot([tx[0], rx[0]], [tx[1], rx[1]], "--", color="tab:blue", alpha=0.7, label="direct path")
ax.scatter(*tx, s=140, marker="^", color="tab:blue", label="TX")
ax.scatter(*rx, s=140, marker="s", color="tab:green", label="RX")
ax.plot(person_path[:, 0], person_path[:, 1], color="tab:orange", linewidth=2, label="person path")

for i, point in enumerate(person_path):
    alpha = 0.25 + 0.75 * (i + 1) / len(person_path)
    ax.add_patch(Circle(point, 0.13, color="tab:red", alpha=alpha))

ax.set_xlim(0, room["width_m"])
ax.set_ylim(0, room["depth_m"])
ax.set_aspect("equal", adjustable="box")
ax.set_xlabel("x position (m)")
ax.set_ylabel("y position (m)")
ax.set_title("Virtual room simulation sketch")
ax.legend(loc="lower right")
plt.show()
```
