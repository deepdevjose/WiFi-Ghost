# 02 Multipath Simulation

Goal: extend the direct/reflected path model into multiple reflected paths with different distances, attenuation, and phase.

```python
import numpy as np
import matplotlib.pyplot as plt

def complex_path(distance_m, wavelength_m, attenuation=1.0):
    phase = -2 * np.pi * distance_m / wavelength_m
    return attenuation * np.exp(1j * phase)
```
