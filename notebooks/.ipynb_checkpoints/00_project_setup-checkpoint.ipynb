# 00 Project Setup

Shared constants and geometry for the WiFiGhost simulation notebooks.

The notebooks are intentionally standalone, but this file defines the baseline assumptions used throughout the sequence.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)

C = 299_792_458
FREQ_HZ = 2.4e9
WAVELENGTH_M = C / FREQ_HZ

room = {"width_m": 5.0, "depth_m": 4.0, "height_m": 2.5}
tx = np.array([0.6, 2.0])
rx = np.array([4.4, 2.0])

sample_rate_hz = 25
n_samples = 500
time_s = np.arange(n_samples) / sample_rate_hz

WAVELENGTH_M
```

```python
person_path = np.column_stack([
    np.linspace(1.0, 4.0, n_samples),
    1.15 + 0.7 * np.sin(2 * np.pi * time_s / time_s[-1]),
])

fig, ax = plt.subplots(figsize=(7, 5))
ax.plot([0, room["width_m"], room["width_m"], 0, 0], [0, 0, room["depth_m"], room["depth_m"], 0], color="0.2")
ax.scatter(*tx, s=120, marker="^", label="TX")
ax.scatter(*rx, s=120, marker="s", label="RX")
ax.plot(person_path[:, 0], person_path[:, 1], color="tab:orange", label="person path")
ax.set_aspect("equal", adjustable="box")
ax.set_xlabel("x position (m)")
ax.set_ylabel("y position (m)")
ax.set_title("Shared simulation geometry")
ax.legend(loc="upper right")
plt.show()
```
