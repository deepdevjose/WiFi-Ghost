# 06 Zone Estimation

Estimate front/back/center/left/right zones from signal patterns.

This is coarse classification, not precise localization.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)
zones = ["front", "back", "center", "left", "right"]
n_subcarriers = 56
subcarrier_axis = np.linspace(-1, 1, n_subcarriers)

templates = {
    "front": 0.4 + 0.08 * np.exp(-((subcarrier_axis - 0.45) ** 2) / 0.08),
    "back": 0.4 + 0.08 * np.exp(-((subcarrier_axis + 0.45) ** 2) / 0.08),
    "center": 0.4 + 0.10 * np.exp(-(subcarrier_axis ** 2) / 0.12),
    "left": 0.4 + 0.06 * np.sin(2 * np.pi * subcarrier_axis),
    "right": 0.4 - 0.06 * np.sin(2 * np.pi * subcarrier_axis),
}

def classify(sample):
    errors = {zone: np.linalg.norm(sample - template) for zone, template in templates.items()}
    best_zone = min(errors, key=errors.get)
    confidence = 1.0 / (1.0 + errors[best_zone])
    return best_zone, confidence, errors
```

```python
true_zone = "center"
sample = templates[true_zone] + 0.025 * rng.normal(size=n_subcarriers)
predicted_zone, confidence, errors = classify(sample)

predicted_zone, confidence, errors
```

```python
fig, axes = plt.subplots(2, 1, figsize=(10, 8))
for zone, template in templates.items():
    axes[0].plot(template, label=zone)
axes[0].plot(sample, color="black", linewidth=2.5, label="sample")
axes[0].set_title(f"Predicted zone: {predicted_zone} confidence={confidence:.2f}")
axes[0].set_ylabel("amplitude pattern")
axes[0].legend(ncol=3)

axes[1].bar(errors.keys(), errors.values(), color="tab:blue")
axes[1].set_ylabel("template distance")
axes[1].set_title("Lower distance is better")

fig.tight_layout()
plt.show()
```
