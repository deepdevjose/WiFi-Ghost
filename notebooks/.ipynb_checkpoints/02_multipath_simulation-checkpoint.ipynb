# 02 Multipath Simulation

Direct path plus reflected paths with a moving person effect.

This notebook introduces complex path summation, so amplitude changes come from constructive and destructive interference.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)
C = 299_792_458
FREQ_HZ = 2.4e9
WAVELENGTH_M = C / FREQ_HZ

tx = np.array([0.6, 2.0])
rx = np.array([4.4, 2.0])
room = {"width_m": 5.0, "depth_m": 4.0}

sample_rate_hz = 25
n_samples = 500
time_s = np.arange(n_samples) / sample_rate_hz
person = np.column_stack([
    np.linspace(1.0, 4.0, n_samples),
    1.15 + 0.7 * np.sin(2 * np.pi * time_s / time_s[-1]),
])

def distance(a, b):
    return np.linalg.norm(a - b, axis=-1)

def attenuation(distance_m, scale=1.0):
    return scale / np.maximum(distance_m, 1.0)

def complex_path(distance_m, amp):
    phase = -2 * np.pi * distance_m / WAVELENGTH_M
    return amp * np.exp(1j * phase)
```

```python
direct_distance = distance(tx, rx)
direct = complex_path(direct_distance, attenuation(direct_distance, 1.0))

person_reflection_distance = distance(tx, person) + distance(person, rx)
person_reflection = complex_path(person_reflection_distance, attenuation(person_reflection_distance, 0.35))

# Two weak static wall-like paths to make the channel less toy-like.
static_reflector_a = np.array([2.5, 0.15])
static_reflector_b = np.array([2.5, 3.85])
static_distance_a = distance(tx, static_reflector_a) + distance(static_reflector_a, rx)
static_distance_b = distance(tx, static_reflector_b) + distance(static_reflector_b, rx)
static_paths = (
    complex_path(static_distance_a, attenuation(static_distance_a, 0.18))
    + complex_path(static_distance_b, attenuation(static_distance_b, 0.16))
)

noise = 0.01 * (rng.normal(size=n_samples) + 1j * rng.normal(size=n_samples))
received = direct + static_paths + person_reflection + noise
rssi_like_db = 10 * np.log10(np.abs(received) ** 2 + 1e-12)

rssi_like_db.min(), rssi_like_db.max()
```

```python
fig, axes = plt.subplots(3, 1, figsize=(10, 8), sharex=True)
axes[0].plot(time_s, person_reflection_distance, color="tab:purple")
axes[0].set_ylabel("path length (m)")
axes[0].set_title("Moving person reflected path")

axes[1].plot(time_s, np.angle(person_reflection), color="tab:orange")
axes[1].set_ylabel("phase (rad)")

axes[2].plot(time_s, rssi_like_db, color="tab:blue")
axes[2].set_ylabel("RSSI-like (dB)")
axes[2].set_xlabel("time (s)")

fig.tight_layout()
plt.show()
```
