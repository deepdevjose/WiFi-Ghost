# Wi-Fi CSI Motion Simulation

Simulate Wi-Fi CSI-like motion detection before hardware is ready.

This notebook uses a 2D room model to generate synthetic RSSI, CSI-like amplitude traces, slow environmental drift, and motion scores. The 3D presentation layer is reserved for the React dashboard later.

## Notebook Structure

1. Imports and configuration
2. Room model
3. Transmitter and receiver positions
4. Human movement path
5. Synthetic RSSI model
6. Synthetic CSI amplitude model
7. Temperature/humidity drift
8. Motion score calculation
9. Dynamic plots
10. Export JSON samples for dashboard

```python
import json
from dataclasses import dataclass

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import plotly.graph_objects as go

plt.style.use('seaborn-v0_8-whitegrid')
np.random.seed(7)

NUM_STEPS = 248
DT_SECONDS = 0.5
TIME = np.arange(NUM_STEPS) * DT_SECONDS
CSI_BINS = 30
ROOM_WIDTH_M = 5.0
ROOM_HEIGHT_M = 4.0

@dataclass(frozen=True)
class Point:
    x: float
    y: float
```

```python
room = {
    'width_m': ROOM_WIDTH_M,
    'height_m': ROOM_HEIGHT_M,
    'walls': [
        ((0.0, 0.0), (ROOM_WIDTH_M, 0.0)),
        ((ROOM_WIDTH_M, 0.0), (ROOM_WIDTH_M, ROOM_HEIGHT_M)),
        ((ROOM_WIDTH_M, ROOM_HEIGHT_M), (0.0, ROOM_HEIGHT_M)),
        ((0.0, ROOM_HEIGHT_M), (0.0, 0.0)),
    ],
    'obstacles': [
        {'name': 'table', 'x': 2.0, 'y': 1.2, 'w': 1.0, 'h': 0.6},
        {'name': 'cabinet', 'x': 3.4, 'y': 2.5, 'w': 0.8, 'h': 1.0},
    ],
}

transmitter = Point(0.6, 0.6)
receiver = Point(4.4, 3.4)

room
```

```python
def human_path(t: np.ndarray) -> pd.DataFrame:
    progress = t / t.max()
    x = 1.0 + 3.0 * progress
    y = 1.0 + 0.5 * np.sin(2 * np.pi * progress * 1.5)
    x[progress < 0.18] = 0.9
    y[progress < 0.18] = 1.0
    x[progress > 0.82] = 4.0
    y[progress > 0.82] = 2.9
    return pd.DataFrame({'time_s': t, 'human_x': x, 'human_y': y})

human = human_path(TIME)
human.head()
```

```python
def euclidean(a: Point, b: Point) -> float:
    return float(np.hypot(a.x - b.x, a.y - b.y))


def distance_point_to_segment(px: float, py: float, ax: float, ay: float, bx: float, by: float) -> float:
    abx = bx - ax
    aby = by - ay
    apx = px - ax
    apy = py - ay
    ab_len_sq = abx * abx + aby * aby
    if ab_len_sq == 0:
        return float(np.hypot(apx, apy))
    t = max(0.0, min(1.0, (apx * abx + apy * aby) / ab_len_sq))
    closest_x = ax + t * abx
    closest_y = ay + t * aby
    return float(np.hypot(px - closest_x, py - closest_y))


def synthetic_rssi(tx: Point, rx: Point, human_df: pd.DataFrame) -> np.ndarray:
    base_distance = euclidean(tx, rx)
    path_loss = -34.0 - 18.0 * np.log10(base_distance)
    line_segment = (tx.x, tx.y, rx.x, rx.y)
    human_distance = np.array([
        distance_point_to_segment(x, y, *line_segment) for x, y in zip(human_df['human_x'], human_df['human_y'])
    ])
    motion_attenuation = -7.0 * np.exp(-human_distance / 0.75)
    temporal_fade = 1.2 * np.sin(2 * np.pi * TIME / 18.0)
    noise = np.random.normal(0.0, 0.8, size=TIME.shape)
    return path_loss + motion_attenuation + temporal_fade + noise

rssi = synthetic_rssi(transmitter, receiver, human)

pd.DataFrame({'time_s': TIME[:5], 'rssi_dbm': rssi[:5]})
```

```python
def synthetic_csi_amplitude(tx: Point, rx: Point, human_df: pd.DataFrame, rssi_dbm: np.ndarray) -> np.ndarray:
    subcarrier_index = np.linspace(-1.0, 1.0, CSI_BINS)
    base_channel = 18.0 + 2.5 * np.cos(np.pi * subcarrier_index)
    human_to_tx = np.hypot(human_df['human_x'].to_numpy() - tx.x, human_df['human_y'].to_numpy() - tx.y)
    human_to_rx = np.hypot(human_df['human_x'].to_numpy() - rx.x, human_df['human_y'].to_numpy() - rx.y)
    motion_strength = np.exp(-np.minimum(human_to_tx, human_to_rx) / 1.3)
    phase_swing = 2.0 * np.sin(2 * np.pi * TIME / 7.5)

    csi = []
    for i, _ in enumerate(TIME):
        spectral_warp = 1.6 * motion_strength[i] * np.sin(2 * np.pi * subcarrier_index * 1.8 + phase_swing[i])
        packet_noise = np.random.normal(0.0, 0.5, size=CSI_BINS)
        amplitude = base_channel + spectral_warp + 0.14 * (rssi_dbm[i] - np.mean(rssi_dbm)) + packet_noise
        csi.append(np.clip(amplitude, 0.5, None))
    return np.array(csi)

csi_amplitude = synthetic_csi_amplitude(transmitter, receiver, human, rssi)

csi_amplitude.shape
```

```python
temperature_c = 24.8 + 0.35 * np.sin(2 * np.pi * TIME / 60.0) + 0.15 * np.cos(2 * np.pi * TIME / 140.0)
humidity_percent = 56.0 + 2.5 * np.cos(2 * np.pi * TIME / 75.0) + 0.8 * np.sin(2 * np.pi * TIME / 33.0)

drift_db = 0.28 * (temperature_c - temperature_c.mean()) - 0.11 * (humidity_percent - humidity_percent.mean())

csi_amplitude_drifted = csi_amplitude + drift_db[:, None]

pd.DataFrame({
    'time_s': TIME[:5],
    'temperature_c': temperature_c[:5],
    'humidity_percent': humidity_percent[:5],
    'drift_db': drift_db[:5],
})
```

```python
amplitude_mean = csi_amplitude_drifted.mean(axis=1)
amplitude_std = csi_amplitude_drifted.std(axis=1)
phase_proxy = np.unwrap(np.angle(np.exp(1j * (amplitude_mean / 6.0))))

motion_raw = (
    0.55 * pd.Series(amplitude_std).rolling(5, min_periods=1).mean().to_numpy() +
    0.30 * np.abs(np.gradient(amplitude_mean)) +
    0.15 * np.abs(np.gradient(rssi))
)
motion_score = (motion_raw - motion_raw.min()) / (motion_raw.max() - motion_raw.min() + 1e-9)
state = np.where(motion_score > 0.58, 'motion', np.where(motion_score > 0.32, 'transition', 'static'))

features = pd.DataFrame({
    'time_s': TIME,
    'rssi_dbm': rssi,
    'amplitude_mean': amplitude_mean,
    'amplitude_std': amplitude_std,
    'temperature_c': temperature_c,
    'humidity_percent': humidity_percent,
    'motion_score': motion_score,
    'state': state,
})

features.head()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

ax = axes[0, 0]
ax.set_title('2D Room Layout')
ax.set_xlim(0, ROOM_WIDTH_M)
ax.set_ylim(0, ROOM_HEIGHT_M)
ax.set_xlabel('x (m)')
ax.set_ylabel('y (m)')
for (x1, y1), (x2, y2) in room['walls']:
    ax.plot([x1, x2], [y1, y2], color='black', linewidth=2)
for obstacle in room['obstacles']:
    rect = plt.Rectangle((obstacle['x'], obstacle['y']), obstacle['w'], obstacle['h'], alpha=0.25, color='tab:gray')
    ax.add_patch(rect)
ax.scatter([transmitter.x], [transmitter.y], color='tab:red', s=120, label='TX')
ax.scatter([receiver.x], [receiver.y], color='tab:blue', s=120, label='RX')
ax.plot(human['human_x'], human['human_y'], color='tab:green', linewidth=2, label='Human path')
ax.legend(loc='lower left')

axes[0, 1].plot(TIME, rssi, color='tab:purple')
axes[0, 1].set_title('Synthetic RSSI')
axes[0, 1].set_xlabel('Time (s)')
axes[0, 1].set_ylabel('RSSI (dBm)')

axes[1, 0].plot(TIME, amplitude_mean, label='Mean CSI amplitude', color='tab:orange')
axes[1, 0].plot(TIME, amplitude_std, label='CSI amplitude std', color='tab:brown', alpha=0.8)
axes[1, 0].set_title('CSI Summary Features')
axes[1, 0].set_xlabel('Time (s)')
axes[1, 0].legend()

axes[1, 1].plot(TIME, motion_score, color='tab:green')
axes[1, 1].set_title('Motion Score')
axes[1, 1].set_xlabel('Time (s)')
axes[1, 1].set_ylabel('0 to 1')

plt.tight_layout()
plt.show()

heatmap = go.Figure(data=go.Heatmap(z=csi_amplitude_drifted.T, x=TIME, colorscale='Viridis'))
heatmap.update_layout(title='CSI-like Amplitude Heatmap', xaxis_title='Time (s)', yaxis_title='Subcarrier Index')
heatmap
```

```python
export_df = features.copy()
export_df['csi_amplitude'] = [row.tolist() for row in csi_amplitude_drifted]
export_df['csi_phase'] = [np.angle(np.exp(1j * row / 12.0)).tolist() for row in csi_amplitude_drifted]
export_df['sequence_number'] = np.arange(NUM_STEPS)
export_df['device_id'] = 'heltec-rx-01'

sample_columns = [
    'time_s', 'sequence_number', 'device_id', 'rssi_dbm', 'temperature_c',
    'humidity_percent', 'motion_score', 'state', 'csi_amplitude', 'csi_phase'
]
sample_records = export_df[sample_columns].head(12).rename(columns={'time_s': 'timestamp_s'}).to_dict(orient='records')

output_path = 'wifighost_motion_samples.json'
with open(output_path, 'w', encoding='utf-8') as file_handle:
    json.dump(sample_records, file_handle, indent=2)

print(f'Wrote {len(sample_records)} dashboard-ready samples to {output_path}')
sample_records[:2]
```
